# UNet Baseline
Train a baseline UNet on processed breast cancer segmentation data.

```python
import os
import torch
from src.models.unet import UNet
from src.datasets.breast_dataset import build_dataloaders
from src.training.train import train_segmentation
from src.training.evaluate import evaluate_segmentation
```

```python
DATA_ROOT = 'data'
IMG_SIZE = 256
BATCH_SIZE = 4
EPOCHS = 10
LR = 1e-3
SAVE_PATH = os.path.join('models', 'unet_baseline.pth')
```

```python
train_loader = build_dataloaders(DATA_ROOT, 'train', BATCH_SIZE, img_size=IMG_SIZE)
val_loader = build_dataloaders(DATA_ROOT, 'val', BATCH_SIZE, img_size=IMG_SIZE)
```

```python
model = UNet(in_channels=3, n_classes=1, base_ch=32)
model = train_segmentation(model, train_loader, val_loader, epochs=EPOCHS, lr=LR, save_path=SAVE_PATH)
```

```python
metrics = evaluate_segmentation(model, val_loader)
print(metrics)
```
